# Target Variable Distribution

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x='Revenue',
    palette='Set2'
)

plt.title("Purchase Intention Distribution")

plt.xlabel("Revenue")

plt.ylabel("Count")

plt.show()
```

# Purchase Percentage

```python
purchase = df['Revenue'].value_counts(normalize=True)*100

purchase
```

# Pie Chart

```python
purchase.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)

plt.ylabel("")

plt.title("Purchase Percentage")

plt.show()
```

# Numerical Feature Distribution

```python
df[numerical_columns].hist(
    figsize=(18,15),
    bins=20
)

plt.tight_layout()

plt.show()
```

# Correlation Matrix

```python
plt.figure(figsize=(16,10))

corr = df[numerical_columns].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

# Features Correlated with Revenue

```python
corr["Revenue"].sort_values(ascending=False)
```

# Top Correlated Features

```python
corr["Revenue"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Features Correlated with Revenue")

plt.show()
```

# Visitor Type Analysis

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='VisitorType',
    hue='Revenue',
    palette='Set2'
)

plt.title("Visitor Type vs Purchase Intention")
plt.xlabel("Visitor Type")
plt.ylabel("Number of Visitors")

plt.legend(title="Revenue")
plt.show()
```

# Visitor Type Purchase Percentage

```python
visitor_purchase = pd.crosstab(
    df['VisitorType'],
    df['Revenue'],
    normalize='index'
)*100

visitor_purchase.round(2)
```
