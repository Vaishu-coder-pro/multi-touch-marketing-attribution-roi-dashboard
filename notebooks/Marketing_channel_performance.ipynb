```python
# Count campaigns by marketing channel
df['Channel_Used'].value_counts()
```

```python
df['Acquisition_Cost'] = df['Acquisition_Cost'].replace('[$,]', '', regex=True).astype(float)

channel_summary = (
    df.groupby('Channel_Used')
      .agg(
          Total_Campaigns=('Campaign_ID', 'count'),
          Total_Clicks=('Clicks', 'sum'),
          Total_Impressions=('Impressions', 'sum'),
          Avg_Conversion_Rate=('Conversion_Rate', 'mean'),
          Avg_Acquisition_Cost=('Acquisition_Cost', 'mean'),
          Avg_ROI=('ROI', 'mean'),
          Avg_Engagement_Score=('Engagement_Score', 'mean')
      )
      .reset_index()
)

channel_summary
```

```python
channel_summary[['Avg_Conversion_Rate',
                 'Avg_Acquisition_Cost',
                 'Avg_ROI',
                 'Avg_Engagement_Score']] = (
    channel_summary[['Avg_Conversion_Rate',
                     'Avg_Acquisition_Cost',
                     'Avg_ROI',
                     'Avg_Engagement_Score']]
    .round(2)
)

channel_summary
```

```python
channel_summary = channel_summary.sort_values(
    by='Avg_ROI',
    ascending=False
)

channel_summary
```

```python
channel_summary.to_csv(
    "channel_summary.csv",
    index=False
)
```

```python
# Marketing Channel Performance Table

channel_performance = (
    df.groupby('Channel_Used')
      .agg(
          Traffic=('Clicks', 'sum'),
          Audience_Reach=('Impressions', 'sum'),
          Avg_Conversion_Rate=('Conversion_Rate', 'mean'),
          Avg_ROI=('ROI', 'mean'),
          Avg_Acquisition_Cost=('Acquisition_Cost', 'mean'),
          Avg_Engagement_Score=('Engagement_Score', 'mean')
      )
      .reset_index()
)

# Round decimal values
channel_performance[['Avg_Conversion_Rate',
                     'Avg_ROI',
                     'Avg_Acquisition_Cost',
                     'Avg_Engagement_Score']] = (
    channel_performance[['Avg_Conversion_Rate',
                         'Avg_ROI',
                         'Avg_Acquisition_Cost',
                         'Avg_Engagement_Score']].round(2)
)

channel_performance
```

```python
channel_performance.to_csv("channel_performance_summary.csv", index=False)
```

```python
import plotly.graph_objects as go

metrics = [
    'Traffic',
    'Audience_Reach',
    'Avg_Conversion_Rate',
    'Avg_ROI',
    'Avg_Acquisition_Cost',
    'Avg_Engagement_Score'
]

fig = go.Figure()

for channel in channel_performance['Channel_Used']:
    values = channel_performance.loc[
        channel_performance['Channel_Used'] == channel,
        metrics
    ].values.flatten().tolist()

    values.append(values[0])

    fig.add_trace(go.Scatterpolar(
        r=values,
        theta=metrics + [metrics[0]],
        fill='toself',
        name=channel
    ))

fig.update_layout(
    title='Overall Marketing Channel Performance',
    polar=dict(radialaxis=dict(visible=True)),
    showlegend=True
)

fig.show()
```

```python
# Display Channel-wise Performance with Business Interpretation

for _, row in channel_performance.iterrows():

    print("="*80)
    print(f"Marketing Channel : {row['Channel_Used']}")
    print("="*80)

    print(f"Traffic (Clicks)           : {row['Traffic']:,}")
    print(f"Audience Reach             : {row['Audience_Reach']:,}")
    print(f"Average Conversion Rate    : {row['Avg_Conversion_Rate']:.2f}%")
    print(f"Average ROI                : {row['Avg_ROI']:.2f}")
    print(f"Average Acquisition Cost   : {row['Avg_Acquisition_Cost']:.2f}")
    print(f"Average Engagement Score   : {row['Avg_Engagement_Score']:.2f}")

    print("\nBusiness Interpretation:")

    # Traffic
    if row['Traffic'] >= channel_performance['Traffic'].mean():
        print("• High Traffic: This channel attracts a large number of visitors, indicating strong audience interest and effective campaign reach.")
    else:
        print("• Low Traffic: This channel attracts fewer visitors and may require better targeting or increased advertising investment.")

    # Audience Reach
    if row['Audience_Reach'] >= channel_performance['Audience_Reach'].mean():
        print("• High Audience Reach: The channel successfully reaches a broad audience and increases brand visibility.")
    else:
        print("• Low Audience Reach: The campaign exposure is limited and may benefit from expanding its reach.")

    # Conversion Rate
    if row['Avg_Conversion_Rate'] >= channel_performance['Avg_Conversion_Rate'].mean():
        print("• High Conversion Rate: The channel effectively converts visitors into customers.")
    else:
        print("• Low Conversion Rate: Visitors are not converting efficiently. Review landing pages, offers, and audience targeting.")

    # ROI
    if row['Avg_ROI'] >= channel_performance['Avg_ROI'].mean():
        print("• High ROI: Marketing investment is generating strong returns and should be considered for additional budget.")
    else:
        print("• Low ROI: The channel is delivering lower returns. Campaign optimization is recommended.")

    # Acquisition Cost
    if row['Avg_Acquisition_Cost'] <= channel_performance['Avg_Acquisition_Cost'].mean():
        print("• Low Acquisition Cost: Customers are acquired efficiently at a lower cost.")
    else:
        print("• High Acquisition Cost: Customer acquisition is expensive. Review campaign efficiency and targeting.")

    # Engagement
    if row['Avg_Engagement_Score'] >= channel_performance['Avg_Engagement_Score'].mean():
        print("• High Engagement: Customers interact actively with the campaign content.")
    else:
        print("• Low Engagement: Improve content quality and audience engagement strategies.")

    print("\nRecommendation:")
    print("Optimize future marketing strategies based on these performance indicators.")

    print("\n")
```
