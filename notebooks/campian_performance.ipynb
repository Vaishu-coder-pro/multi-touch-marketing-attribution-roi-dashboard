###Total Campaigns

```python
print("Total Campaigns:", df['Campaign_ID'].nunique())
```

###Top 10 Campaigns by ROI

```python
top_clicks = df[['Campaign_ID', 'Clicks']].sort_values(by='Clicks', ascending=False)

top_clicks.head(10)
```

###Top 10 Campaigns by Impressions

```python
top_impressions = df[['Campaign_ID', 'Impressions']].sort_values(by='Impressions', ascending=False)

top_impressions.head(10)
```

###Campaign-wise Conversion Rate

```python
campaign_conversion = df[['Campaign_ID', 'Conversion_Rate']].sort_values(by='Conversion_Rate', ascending=False)

campaign_conversion.head(10)
```

###Campaign-wise Acquisition Cost

```python
campaign_cost = df[['Campaign_ID', 'Acquisition_Cost']].sort_values(by='Acquisition_Cost', ascending=False)

campaign_cost.head(10)
```

###Campaign-wise Engagement Score

```python
campaign_engagement = df[['Campaign_ID', 'Engagement_Score']].sort_values(by='Engagement_Score', ascending=False)

campaign_engagement.head(10)
```

###Campaign Performance Summary

```python
campaign_summary = df.groupby('Campaign_ID').agg({
    'Clicks':'sum',
    'Impressions':'sum',
    'ROI':'mean',
    'Conversion_Rate':'mean',
    'Acquisition_Cost':'sum',
    'Engagement_Score':'mean'
})

campaign_summary.head()
```

###Best Performing Campaign

```python
best_campaign = campaign_summary.sort_values(by='ROI', ascending=False)

best_campaign.head()
```

###Worst Performing Campaign

```python
worst_campaign = campaign_summary.sort_values(by='ROI')

worst_campaign.head()
```

###Campaign ROI

```python
import matplotlib.pyplot as plt

campaign_summary['ROI'].sort_values(ascending=False).head(10).plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Campaigns by ROI")
plt.xlabel("Campaign ID")
plt.ylabel("Average ROI")
plt.xticks(rotation=45)
plt.show()
```

###campaign clicks

```python
campaign_summary['Clicks'].sort_values(ascending=False).head(10).plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Campaigns by Clicks")
plt.xlabel("Campaign ID")
plt.ylabel("Total Clicks")
plt.xticks(rotation=45)
plt.show()
```

###Campaign Conversion Rate

```python
campaign_summary['Conversion_Rate'].sort_values(ascending=False).head(10).plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Campaigns by Conversion Rate")
plt.xlabel("Campaign ID")
plt.ylabel("Conversion Rate")
plt.xticks(rotation=45)
plt.show()
```
