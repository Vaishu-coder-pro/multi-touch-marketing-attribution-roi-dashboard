# Feature Classification

```python
numerical_features = df.select_dtypes(include=['int64','float64']).columns.tolist()

categorical_features = df.select_dtypes(include=['object']).columns.tolist()

datetime_features = []

for col in df.columns:
    if 'date' in col.lower():
        datetime_features.append(col)

identifier_features = [col for col in df.columns if 'id' in col.lower()]

print("="*70)
print("FEATURE CLASSIFICATION")
print("="*70)

print(f"\n Numerical Features ({len(numerical_features)})")
print(numerical_features)

print(f"\n Categorical Features ({len(categorical_features)})")
print(categorical_features)

print(f"\n Datetime Features ({len(datetime_features)})")
print(datetime_features)

print(f"\n Identifier Columns ({len(identifier_features)})")
print(identifier_features)
```

```python
feature_summary = pd.DataFrame({
    "Feature Type":
    ["Numerical",
     "Categorical",
     "Datetime",
     "Identifier"],
    "Count":
    [len(numerical_features),
     len(categorical_features),
     len(datetime_features),
     len(identifier_features)]
})

plt.figure(figsize=(8,5))

sns.barplot(
    data=feature_summary,
    x="Feature Type",
    y="Count",
    palette="viridis"
)

plt.title("Feature Classification",fontsize=16)
plt.xlabel("")
plt.ylabel("Number of Features")

for i,v in enumerate(feature_summary["Count"]):
    plt.text(i,v+0.1,str(v),ha='center',fontsize=12)

plt.show()
```

# Data Quality Score

```python
missing_score = 25

duplicate_score = 25

datatype_score = 20

unique_score = 15

completeness_score = 15


if df.isnull().sum().sum()>0:
    missing_score -=10

if df.duplicated().sum()>0:
    duplicate_score -=10


quality_score = (
missing_score+
duplicate_score+
datatype_score+
unique_score+
completeness_score
)

print("="*70)
print("DATA QUALITY SCORE")
print("="*70)

print(f"Score : {quality_score}/100")

if quality_score>=90:
    print("Excellent Dataset")

elif quality_score>=75:
    print("Good Dataset")

elif quality_score>=60:
    print("Average Dataset")

else:
    print("Poor Dataset")
```

```python
plt.figure(figsize=(7,1.8))

plt.barh(
    ["Quality Score"],
    [100],
    color="lightgray"
)

plt.barh(
    ["Quality Score"],
    [quality_score],
    color="green"
)

plt.xlim(0,100)

plt.title(f"Dataset Quality : {quality_score}/100")

plt.show()
```
