# Missing Values

```python
missing = pd.DataFrame({

"Missing Values":df.isnull().sum(),

"Percentage":round(df.isnull().sum()/len(df)*100,2)

})

missing
```

```python
missing = df.isnull().sum().sort_values(ascending=False)

plt.figure(figsize=(12,5))

missing.plot(kind='bar', color='steelblue')

plt.title("Missing Values by Column", fontsize=16)
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

for i, value in enumerate(missing):
    plt.text(i, value + 0.02, str(value), ha='center', fontsize=9)

plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
if df.isnull().sum().sum() == 0:
    print(" No Missing Values Found in the Dataset.")
else:
    print("Missing Values Exist.")
```

# Encode Categorical Variables

```python
month_map = {

'Feb':2,

'Mar':3,

'May':5,

'June':6,

'Jul':7,

'Aug':8,

'Sep':9,

'Oct':10,

'Nov':11,

'Dec':12

}

df["Month"] = df["Month"].map(month_map)
```

```python
visitor_map = {

"New_Visitor":0,

"Returning_Visitor":1,

"Other":2

}

df["VisitorType"] = df["VisitorType"].map(visitor_map)
```

```python
df["Weekend"] = df["Weekend"].astype(int)
```

```python
df["Revenue"] = df["Revenue"].astype(int)
```

# Data Quality Report

```python
quality = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Missing %": round(df.isnull().mean() * 100, 2),
    "Unique Values": df.nunique()
})

quality["Duplicate Rows"] = df.duplicated().sum()

quality
```

```python
quality["Quality Status"] = np.where(
    quality["Missing Values"] == 0,
    "Good",
    "Needs Cleaning"
)

quality
```

# Basic Feature Engineering

```python
df["Total_Page_Visits"] = (

df["Administrative"]

+

df["Informational"]

+

df["ProductRelated"]

)
```

```python
df["Total_Duration"] = (

df["Administrative_Duration"]

+

df["Informational_Duration"]

+

df["ProductRelated_Duration"]

)
```

```python
df["Engagement_Score"] = (

df["Total_Page_Visits"]

*

df["Total_Duration"]

)
```

```python
df["Product_Interest_Ratio"] = (

df["ProductRelated"]

/

df["Total_Page_Visits"]

)
```

```python
df["Product_Interest_Ratio"] = df["Product_Interest_Ratio"].fillna(0)
```
